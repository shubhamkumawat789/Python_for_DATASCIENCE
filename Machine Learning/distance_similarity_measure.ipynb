## Title: Write a program to compute various distance measures and similarity measure between the samples. 
**What are distance measures?**

 Distance measures quantify the dissimilarity between samples. A smaller distance indicates higher similarity.

```python
import numpy as np
import pandas as pd

data = pd.DataFrame({
    'Feature1': [2, 5],
    'Feature2': [3, 7],
    'Feature3': [4, 9]
})

print("Dataset:\n", data)

x = data.iloc[0].values
y = data.iloc[1].values   
```

### 1. Euclidean Distance
Euclidean distance is considered the traditional metric for problems with geometry. It can be simply explained as the ordinary distance between two points. It is one of the most used algorithms in the cluster analysis.

- **Formula**:  
$
d(x,y) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}
$


```python
distance = np.sqrt(np.sum((x - y) ** 2))
print("\nEuclidean Distance between sample 1 and sample 2:", distance)
```

### 2. Manhattan Distance:
Manhattan Distance determines the absolute difference among the pair of the coordinates. Suppose we have two points P and Q to determine the distance between these points we simply have to calculate the perpendicular distance of the points from X-Axis and Y-Axis. In a plane with P at coordinate (x1, y1) and Q at (x2, y2). Manhattan distance between P and Q = |x1 – x2| + |y1 – y2|

$
d(x,y) = \sum_{i=1}^{n} \lvert x_i - y_i \rvert
$

```python
distance = np.sum(np.abs(x - y))
print("\nManhattan Distance between sample 1 and sample 2:", distance)
```

### 3. Minkowski Distance:
Minkowski distance is a generalized distance measure that includes both `Euclidean` and `Manhattan` distances as special cases, controlled by a parameter p.

$
d(x,y) = \left( \sum_{i=1}^{n} \lvert x_i - y_i \rvert^p \right)^{\tfrac{1}{p}}
$

```python
def minkowski_distance(x, y, p):
    return np.power(np.sum(np.abs(x - y) ** p), 1/p)

p = 3  
distance = minkowski_distance(x, y, p)

print(f"\nMinkowski Distance (p={p}) between sample 1 and sample 2:", distance)
```

### 4. Chebyshev Distance
Maximum absolute difference along any coordinate (L∞ norm). 

$
d(x,y) = \max_i |x_i - y_i|
$



```python
distance = np.max(np.abs(x - y))
print("\nChebyshev Distance between sample 1 and sample 2:", distance)
```

### 5. Hamming distance
Hamming distance measures the **number of positions where two strings (or vectors of equal length) differ**.  


$
d(x,y) = \sum_{i=1}^{n} [x_i \neq y_i]
$


```python
def hamming_distance(x, y, normalize=False):
    x = np.array(x)
    y = np.array(y)

    mismatches = np.sum(x != y)

    if normalize:
        # Divide by length to get a value between 0 and 1
        return mismatches / len(x)
    else:
        return mismatches

dist = hamming_distance(x, y)
norm_dist = hamming_distance(x, y, normalize=True)

print("Hamming Distance:", dist)
print("Normalized Hamming Distance:", norm_dist)
```

## Similarity Measures (numeric / vector data)
**What are similarity measures?**

Similarity measures quantify the similarity between samples. A higher similarity indicates smaller distance.


### Cosine Similarity & Cosine Distance

Cosine similarity measures the **cosine of the angle** between two vectors. It focuses on the **orientation** (direction) rather than the **magnitude** of the vectors.  

**Cosine Similarity:**

$
\text{Cosine}(x,y) = \frac{x \cdot y}{||x|| \; ||y||}
$

**Cosine Distance:**

$
d(x,y) = 1 - \frac{x \cdot y}{||x|| \; ||y||}
$

```python
def cosine_similarity(x, y):
    num = np.dot(x, y)
    den = (np.linalg.norm(x) * np.linalg.norm(y)) + 1e-12 
    return num / den

def cosine_distance(x, y):
    return 1 - cosine_similarity(x, y)

sim = cosine_similarity(x, y)
dist = cosine_distance(x, y)

print("\nCosine Similarity:", sim)
print("Cosine Distance:", dist)
```

### Jaccard distance:
The Jaccard distance is set-based distance that compares dissimilarity by looking at the ratio of unique to common elements.

$
d(A,B) = 1 - \frac{|A \cap B|}{|A \cup B|}
$

```python
def jaccard_distance_sets(a, b):
    A, B = set(a), set(b)
    intersection = len(A & B)
    union = len(A | B)
    return 1 - intersection / (union + 1e-12)

distance = jaccard_distance_sets(x, y)
print("\nJaccard Distance between sample 1 and sample 2:", distance)
```

### Pearson Correlation Coefficient (as Similarity)
 Measures the **linear relationship** between paired measurements across dimensions; essentially cosine similarity of **mean-centered** vectors.

**Formula (Similarity):**
$
r(x,y)
= \frac{\sum_{i=1}^{n}\big(x_i-\bar{x}\big)\big(y_i-\bar{y}\big)}
{\sqrt{\sum_{i=1}^{n}\big(x_i-\bar{x}\big)^2}\;
 \sqrt{\sum_{i=1}^{n}\big(y_i-\bar{y}\big)^2}}
$

```python
def pearson_similarity(x, y):
    x, y = np.asarray(x), np.asarray(y)

    x_mean, y_mean = np.mean(x), np.mean(y)
    x_centered = x - x_mean
    y_centered = y - y_mean

    numerator = np.sum(x_centered * y_centered)

    denominator = np.sqrt(np.sum(x_centered**2)) * np.sqrt(np.sum(y_centered**2)) + 1e-12

    return numerator / denominator

sim = pearson_similarity(x, y)
print("\nPearson Correlation Coefficient (similarity):", sim)
```
