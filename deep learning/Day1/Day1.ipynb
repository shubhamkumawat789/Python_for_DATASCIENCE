# PyTorch Tensors — Basics
### Create Tensors

```python
import torch

x = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)

r = torch.randn(3, 3)

z = torch.zeros(2, 3)
o = torch.ones(3, 2)

print(x, r, z, o)
```

### Check GPU availability

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("using device:", device )
```

### move tensor to GPU

```python
x_gpu = x.to(device)
x_gpu
```

### Tensor operation

```python
a = torch.tensor([1,2, 3.], requires_grad=True)
b = a * 2
c = b.sum()
```

### Autograd

```python
c.backward()
print(a.grad)
```

# TensorFlow Tensors — Basics
### Create Tensors

```python
import tensorflow as tf

x = tf.constant([[1,2], [3,4]], dtype=tf.float32)
r = tf.random.normal((3,3))
z = tf.zeros((2,3))
o = tf.ones((3, 2))

print(x, r, z, o)
```

### Check GPU

```python
print("GPU available:", tf.config.list_physical_devices("GPU"))
```

### Tensor operations

```python
a = tf.Variable([1.0, 2.0, 3.0])
b = a * 2
c = tf.reduce_sum(b)
print(c)
```

### Autograd (tf.GradientTape)

```python
with tf.GradientTape() as tape:
    a = tf.Variable([1.0, 2.0, 3.0])
    b = a * 2 
    c = tf.reduce_sum(b)

grad = tape.gradient(c, a)
print(grad)
```

# Keras Tensors
### Create Tensors

```python
from tensorflow import keras
import tensorflow as tf

x = tf.constant([[5., 6.], [7., 8.]])
print(x)
```

### Autograd in Keras (uses TF GradientTape)

```python
import tensorflow as tf

with tf.GradientTape() as tape:
    w = tf.Variable([1.0, 2.0])
    y = w * 3
grad = tape.gradient(y, w)
print(grad)
```

## Exercise 5 — TensorFlow GradientTape Practice

#### Task: In TensorFlow, compute gradient of
#### f(w) = w^3 + 2w
#### for w = [1.,2.,3.].

```python
import tensorflow as tf

w = tf.Variable([1.,2.,3.])
with tf.GradientTape() as tape:
    f = w**3 + 2*w
grad = tape.gradient(f, w)
print(grad)
```
